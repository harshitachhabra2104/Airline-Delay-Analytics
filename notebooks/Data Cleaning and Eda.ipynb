✈️ Section 4.2: Airline Performance Analysis
📌 Business Question 1
Which airlines operate the highest number of flights?

```python
# ==========================================
# Flights Operated by Each Airline
# ==========================================

airline_flights = (
    flight_df["carrier_name"]
    .value_counts()
    .reset_index()
)

airline_flights.columns = ["Airline", "Total Flights"]

display(airline_flights)

plt.figure(figsize=(12,6))

plt.bar(
    airline_flights["Airline"],
    airline_flights["Total Flights"]
)

plt.title("Flights Operated by Airline")
plt.xlabel("Airline")
plt.ylabel("Number of Flights")

plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

💼 Business Insight
High-volume airlines have greater operational responsibility and require continuous performance monitoring.

```python
# ==========================================
# Departure Delay Rate by Airline
# ==========================================

airline_delay = (
    flight_df
    .groupby("carrier_name")["dep_del15"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

airline_delay["Delay Rate (%)"] = airline_delay["dep_del15"] * 100

display(airline_delay[["carrier_name","Delay Rate (%)"]])

plt.figure(figsize=(12,6))

plt.bar(
    airline_delay["carrier_name"],
    airline_delay["Delay Rate (%)"]
)

plt.title("Departure Delay Rate by Airline")
plt.xlabel("Airline")
plt.ylabel("Delay Rate (%)")

plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

📌 Observation
Some airlines consistently experience higher departure delay rates.
💼 Business Insight
Airlines with higher delay percentages may require operational improvements in scheduling, staffing, or aircraft turnaround.

📌 Business Question 3
Which airlines have the best on-time performance?

```python
# ==========================================
# On-Time Performance by Airline
# ==========================================

airline_ontime = (
    flight_df
    .groupby("carrier_name")["dep_del15"]
    .mean()
)

airline_ontime = (
    (1 - airline_ontime) * 100
).sort_values(ascending=False)

display(airline_ontime.reset_index(name="On-Time Performance (%)"))

plt.figure(figsize=(12,6))

plt.bar(
    airline_ontime.index,
    airline_ontime.values
)

plt.title("On-Time Performance by Airline")
plt.xlabel("Airline")
plt.ylabel("On-Time Performance (%)")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

📌 Observation
Certain airlines consistently maintain higher on-time departure performance.
💼 Business Insight
Strong on-time performance reflects efficient scheduling, resource allocation, and operational planning.

📌 Business Question 4
Rank airlines based on delay performance

```python
# ==========================================
# Airline Performance Ranking
# ==========================================

airline_rank = (
    flight_df
    .groupby("carrier_name")["dep_del15"]
    .mean()
    .sort_values()
    .reset_index()
)

airline_rank["Delay Rate (%)"] = airline_rank["dep_del15"] * 100

airline_rank["Rank"] = range(1, len(airline_rank)+1)

display(
    airline_rank[
        ["Rank","carrier_name","Delay Rate (%)"]
    ]
)
```

💼 Business Insight

📌 Business Question 5
Is there a relationship between flight volume and delay rate?

```python
# ==========================================
# Flight Volume vs Delay Rate
# ==========================================

volume_delay = (
    flight_df
    .groupby("carrier_name")
    .agg(
        Total_Flights=("carrier_name","count"),
        Delay_Rate=("dep_del15","mean")
    )
    .reset_index()
)

volume_delay["Delay Rate (%)"] = volume_delay["Delay_Rate"] * 100

display(volume_delay)

plt.figure(figsize=(8,6))

plt.scatter(
    volume_delay["Total_Flights"],
    volume_delay["Delay Rate (%)"]
)

plt.title("Flight Volume vs Delay Rate")
plt.xlabel("Total Flights")
plt.ylabel("Delay Rate (%)")

plt.grid(True)

plt.show()
```

Observation
High flight volume does not necessarily result in higher delay rates.
💼 Business Insight
Operational efficiency depends on scheduling, resource management, and airport conditions rather than flight volume alone.

📌 Business Question 1
Does aircraft age influence departure delays?

```python
# ==========================================
# Plane Age vs Departure Delay
# ==========================================

plane_delay = (
    flight_df
    .groupby("plane_age")["dep_del15"]
    .mean()
    .reset_index()
)

plane_delay["Delay Rate (%)"] = plane_delay["dep_del15"] * 100

display(plane_delay.head())

plt.figure(figsize=(10,5))

plt.plot(
    plane_delay["plane_age"],
    plane_delay["Delay Rate (%)"],
    marker="o"
)

plt.title("Departure Delay Rate by Aircraft Age")
plt.xlabel("Aircraft Age (Years)")
plt.ylabel("Delay Rate (%)")
plt.grid(True)

plt.show()
```

📌 Observation

Aircraft age may show varying delay rates across different age groups.

💼 Business Insight

Older aircraft may require more maintenance, potentially affecting operational efficiency.

📌 Business Question 2
Does flight distance affect departure delays?

```python
# ==========================================
# Distance Group vs Delay Rate
# ==========================================

distance_delay = (
    flight_df
    .groupby("distance_group")["dep_del15"]
    .mean()
    .reset_index()
)

distance_delay["Delay Rate (%)"] = distance_delay["dep_del15"] * 100

display(distance_delay)

plt.figure(figsize=(10,5))

plt.bar(
    distance_delay["distance_group"].astype(str),
    distance_delay["Delay Rate (%)"]
)

plt.title("Departure Delay Rate by Distance Group")
plt.xlabel("Distance Group")
plt.ylabel("Delay Rate (%)")

plt.show()
```
